```python
import pandas as pd
```

```python
dataset=pd.read_csv("Placement.csv")
```

```python
#Scatter Plot
import seaborn as sns
import matplotlib.pyplot as plt

data=dataset

sns.scatterplot(x='mba_p',y='salary',hue='gender',data=data)
```

```python
#Line Plot
import seaborn as sns
import matplotlib.pyplot as plt

data=dataset

sns.lineplot(x='mba_p',y='salary',hue='gender',data=data)
plt.show()
```

```python
#Bar plot
import seaborn as sns
import matplotlib.pyplot as plt

data=dataset

sns.barplot(x='degree_t',y='salary',hue='gender',data=data)
plt.show()
```

```python
#Box Plot
import pandas as pd
import matplotlib.pyplot as plt
data=dataset
sns.boxplot(x='specialisation',y='salary',hue='gender',data=data)
plt.show()
```

```python
#5 HistPlot
import seaborn as sns
import matplotlib.pyplot as plt
data=dataset
sns.histplot(data['salary'],kde=True)
plt.show()
```

```python
#6.Heat map
import seaborn as sns
import matplotlib.pyplot as plt
data=dataset
corr=data.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```python
#PairPlot
import seaborn as sns
import matplotlib.pyplot as plt
data=dataset
sns.pairplot(data,hue='degree_t')
plt.show()
```

```python
#Swarm Plot
import seaborn as sns
import matplotlib.pyplot as plt
data=dataset
sns.swarmplot(x="degree_t",y="degree_p",data=data,hue="gender")
plt.show()
```

```python
#swarmplot overlaying boxplot
import seaborn as sns
import matplotlib.pyplot as plt
data=dataset
sns.boxplot(x="degree_t",y="degree_p",data=data,hue="gender")
sns.swarmplot(x="degree_t",y="degree_p",data=data,hue="gender")
plt.show()
```

```python
#pointplot
import seaborn as sns
import matplotlib.pyplot as plt
data=dataset
sns.pointplot(x="degree_t",y="degree_p",data=dataset,hue="gender")
plt.show()
```
