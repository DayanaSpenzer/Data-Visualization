```python
raw_dataset['RATING_TYPE'].value_counts()
```

```python
raw_dataset['RATING_TYPE'].replace(to_replace='Excelente' , value='Excellent', inplace=True) 
raw_dataset['RATING_TYPE'].replace(to_replace=['Veľmi dobré','Bardzo dobrze','Muy Bueno','Velmi dobré'] , value='Very Good',inplace=True)
raw_dataset['RATING_TYPE'].replace(to_replace=['Skvělá volba','Dobrze','Bueno','Buono','Dobré','Bom','Skvělé'] , value='Good', inplace=True)
raw_dataset['RATING_TYPE'].replace(to_replace=['Priemer','Média','Çok iyi'] , value='Average', inplace=True) 
raw_dataset['RATING_TYPE'].replace(to_replace=['Průměr','Promedio','Ortalama','Muito Bom','İyi'] , value='Poor', inplace=True)
raw_dataset['RATING_TYPE'].replace(to_replace=['Baik','Biasa','Media','Sangat Baik'] , value='Very Poor', inplace=True) 
```

```python
raw_dataset['RATING_TYPE'].value_counts()
```

```python
raw_dataset['REGION']
```

```python
#Removing the irrelevant text from the region column
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('[a-zA-Z].+-- ','',regex=True)
```

```python
 # Removing the West & East from the Region column 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace(' West| west| East| east','',regex=True)
```

```python
raw_dataset['REGION'].value_counts()
```

```python
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('4 Bungalows|7 Andheri|Azad Nagar|Near Andheri Station|Veera Desai Area','Azad Nagar',regex=True)
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Bandra Kurla Complex','Bandra',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('CBD-Belapur','CBD Belapur',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Girgaon Chowpatty','Chowpatty',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Dadar Shivaji Park','Dadar',regex=True)
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Flea Bazaar Café|Kamala Mills Compound','Lower Parel',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Runwal Green','Mulund',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Mumbai CST Area','Mumbai Central',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Kopar Khairane|Seawoods|Turbhe|Ulwe','Navi Mumbai',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('New Panvel|Old Panvel','Panvel',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Kamothe','Sion',regex=True) 
raw_dataset['REGION'] = raw_dataset['REGION'].str.replace('Ghodbunder Road|Majiwada','Thane',regex=True)
```

```python
#2.How to Categorize the restaurants in Mumbai with the rating type?
```

```python
rating_type_df = zomato_dataset['RATING_TYPE'].value_counts().reset_index() 
rating_type_df.rename(columns={'index':'RATING TYPE', 'RATING_TYPE':'RATING CATEGORY'}, inplace=True) 
rating_type_df 
```

```python
fig = px.pie(rating_type_df, names='RATING TYPE', values='COUNT OF RESTAURANTS', color='RATING TYPE',  
             title='Percentage of Restaurants by Rating Type').update_traces(textposition='inside', textinfo='percent+label')
fig.show() 
```
